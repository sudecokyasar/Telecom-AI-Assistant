```python
import pandas as pd
import numpy as np

# Veriyi oku
df_sample = pd.read_csv('../data/raw/sample.csv')

print(f"Toplam Satır Sayısı: {df_sample.shape[0]}")
print(f"Toplam Sütun Sayısı: {df_sample.shape[1]}")
df_sample.head()
```

```python
# Sütun tipleri ve eksik değer analizi
df_sample.info()
print("\nEksik Değer Sayıları:")
print(df_sample.isnull().sum())
```

```python
# 'inbound' sütunu: True ise müşteri sorusu, False ise şirket/destek yanıtı
print("Inbound Dağılımı:")
print(df_sample['inbound'].value_counts())

# En çok destek veren hesapları listele (Şirketler inbound=False olanlardır)
top_authors = df_sample[df_sample['inbound'] == False]['author_id'].value_counts()
print("\nEn Aktif Destek Hesapları:")
print(top_authors.head(10))
```

```python
# Müşterilerin yazdığı ilk 5 tweet metni
customer_tweets = df_sample[df_sample['inbound'] == True]['text']

for idx, text in enumerate(customer_tweets.head(5), 1):
    print(f"Örnek {idx}: {text}\n{'-'*50}")
```
